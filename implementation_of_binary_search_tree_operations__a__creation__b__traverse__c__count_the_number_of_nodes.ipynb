```python
class Node:
  #constructor to create a new node
  def __init__(self, key):
    self.data = key
    self.left = None
    self.right = None

def insert(node, data):
  if node is None:
    return Node(data)
  if data <= node.data:
    node.left = insert(node.left, data)
  else:
    node.right = insert(node.right, data)
  return node

def minValue(node):
  current = node
  while current.left is not None:
    current = current.left
  return current.data

root = None
root = insert(root, 4)
insert(root, 2)
insert(root, 1)
insert(root, 3)
insert(root, 6)
insert(root, 5)
print("\nMinimum value in BST is %d" % (minValue(root)))
```

```python
INT_MIN = -2**32
def canRepresentBST(pre):
  s = []
  root = INT_MIN
  for value in pre:
    # If we find a value smaller than the current root, then it's not a valid BST preorder
    if value < root:
      return False

    # While stack is not empty and the top of stack is smaller than current value,
    # pop elements from stack. The last popped element becomes the new root (lower bound)
    # for subsequent right children.
    while(len(s) > 0 and s[-1] < value):
      root = s.pop()

    # Push current value to stack
    s.append(value)

  return True

pre1 = [40 , 30 , 35 , 80 , 100]
print("true" if canRepresentBST(pre1) == True else "false")
pre2 = [40 , 30 , 35 , 20 , 80 , 100]
print("true" if canRepresentBST(pre2) == True else "false")
```

```python
class newNode:
  def __init__(self, data):
    self.data = data
    self.left = None
    self.right = None

  @staticmethod
  def getCount(root, low, high):
    # Base case
    if root is None:
      return 0

    # If current node is in range, then include it in count and recur for left and right children
    if root.data >= low and root.data <= high:
      return (1 + newNode.getCount(root.left, low, high) + newNode.getCount(root.right, low, high))

    # If current node is smaller than low, then process only right child
    elif root.data < low:
      return newNode.getCount(root.right, low, high)

    # If current node is greater than high, then process only left child
    else: # root.data > high
      return newNode.getCount(root.left, low, high)

if __name__ == '__main__':
    root = newNode(10)
    root.left = newNode(5)
    root.right = newNode(50)
    root.left.left = newNode(1)
    root.right.left = newNode(40)
    root.right.right = newNode(100)

    l = 5
    h = 45
    print("Count of nodes between [", l, ", ", h,"] is ", newNode.getCount(root, l, h))
```
